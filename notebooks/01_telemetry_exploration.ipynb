```python
fig, ax1 = plt.subplots()

ax1.plot(telemetry["time"], telemetry["speed"])
ax1.set_xlabel("times (s)")
ax1.set_ylabel("Speed (km/h)")

ax2 = ax1.twinx()

ax2.plot(telemetry["time"], telemetry["brake"])
ax2.set_ylabel("Brake Input")

plt.title("Speed and Braking")
plt.show()
```
